```python
# 1. Slice your DF (You already did this)
all_lvl0_cols = df.columns.get_level_values(0).unique()
pos = all_lvl0_cols.get_loc('전기 \n배율')
cols_to_keep = all_lvl0_cols[:pos]
df = df[cols_to_keep]

# 2. Map the old names to the new names using a dictionary
# This ensures every column is renamed correctly, regardless of position
name_map = {old: old.replace(' ', '').replace('\n', '') for old in cols_to_keep}

# 3. Reconstruct the MultiIndex
# col[0] is Level 0, col[1:] are the remaining levels
df.columns = pd.MultiIndex.from_tuples(
    [(name_map[col[0]], *col[1:]) for col in df.columns]
)

print(df.columns.get_level_values(0).unique())

print(df.columns)
```

```python
import numpy as np

# def insert_columns(df, target, pct_name, pct_series):
    
def create_change_columns(df):
    # Mapping: (Previous, Current, Insert_Target, Change_Name, Percent_Name)
    tasks = [
        ('water_previous', 'water_current', 'water_usage_m3', 'water_change', 'water_pct'),
        ('hwater_previous', 'hwater_current', 'hwater_usage_m3', 'hwater_change', 'hwater_pct'),
        ('elect_previous', 'elect_current', 'elect_usage_kw', 'elect_change', 'elect_pct'),
        ('heat_previous', 'heat_current', 'heat_usage_m3_mwh', 'heat_change', 'heat_pct')
    ]
    
    for prev, curr, target, diff_name, pct_name in tasks:
        if prev in df.columns and curr in df.columns:
            # 1. Clean and convert
            v_curr = pd.to_numeric(df[curr].astype(str).str.replace(',', ''), errors='coerce')
            v_prev = pd.to_numeric(df[prev].astype(str).str.replace(',', ''), errors='coerce')
            
            # 2. Quantitative Change (Current - Previous)
            diff_series = v_curr - v_prev
            
            # 3. Proportional Change (Percentage)
            # .div() with fill_value=0 prevents errors if previous is 0
            pct_series = (diff_series.div(v_prev.replace(0, np.nan)) * 100).round(2)


            # 4. Insert both (Insert Pct first, then Change so Change is closer to target)
            pos = df.columns.get_loc(target)
            df.insert(loc=pos + 1, column=diff_name, value=diff_series)
            df.insert(loc=pos + 2, column=pct_name, value=pct_series)
            
            # df = insert_column(df, target, pct_name, pct_series)
            # df = insert_column(df, target, diff_name, diff_series)
    return df

df_all = df_drop.copy()
df_all = create_change_columns(df_all)
display(df_all.head())
df_all['water_pct'].describe()# display(df_d.head())
```

```python
import matplotlib.pyplot as plt

def plot_utility_histograms_with_extremes(df, change_col, pct_col, title="Utility Distribution", threshold=0.20):
    # Create a figure with 2 subplots
    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(14, 5))
    
    # Define columns to plot and their settings
    plot_configs = [
        {'col': change_col, 'ax': axes[0], 'color': 'skyblue', 'label': 'Change'},
        {'col': pct_col, 'ax': axes[1], 'color': 'salmon', 'label': 'Pct (%)'}
    ]

    for config in plot_configs:
        col_data = df[config['col']].dropna()
        ax = config['ax']
        
        # 1. Plot Histogram
        ax.hist(col_data, bins=30, color=config['color'], edgecolor='black', alpha=0.7)
        
        # 2. Calculate and Plot Percentile Thresholds
        low_val = col_data.quantile(threshold)
        high_val = col_data.quantile(1 - threshold)
        
        ax.axvline(low_val, color='blue', linestyle='dashed', linewidth=2, label=f'Bottom {threshold*100:.0f}%')
        ax.axvline(high_val, color='red', linestyle='dashed', linewidth=2, label=f'Top {threshold*100:.0f}%')
        
        # Formatting
        ax.set_title(f"{config['label']}: {config['col']}")
        ax.set_xlabel('Value')
        ax.set_ylabel('Frequency')
        ax.legend()

    plt.suptitle(title, fontsize=16)
    plt.tight_layout()
    plt.show()

# Execution for 2026 Analysis
plot_utility_histograms_with_extremes(df_a, 'water_change', 'water_pct', title="Water Usage Analysis (2026)")
plot_utility_histograms_with_extremes(df_b, 'hwater_change', 'hwater_pct', title="Hot Water Usage Analysis (2026)")
plot_utility_histograms_with_extremes(df_c, 'elect_change', 'elect_pct', title="Electricity Usage Analysis (2026)")
plot_utility_histograms_with_extremes(df_d, 'heat_change', 'heat_pct', title="Heat Usage Analysis (2026)")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def interactive_overlap_dashboard_v2(
    df_all: pd.DataFrame,
    df_a: pd.DataFrame,
    df_b: pd.DataFrame,
    df_c: pd.DataFrame,
    df_d: pd.DataFrame,
    building_col: str = "building",
    brand_col: str = "brand",
    default_bins: int = 50,
    default_thr_change=(0.20, 0.80),
    default_thr_pct=(0.20, 0.80),
    max_rows_show: int = 200,
):
    """
    DF chooser + Category chooser (prefix inferred from *_change + *_pct columns)

    Shows:
      - histograms with tails shaded (change, pct)
      - quantile bounds summary

      - Change tails (ROWS):
          * Change LOW  (change < lo_c)
          * Change HIGH (change > hi_c)

      - Pct tails (ROWS):
          * Pct LOW  (pct < lo_p)
          * Pct HIGH (pct > hi_p)

      - Overlap quadrants (ROWS):
          1) change low  & pct low
          2) change low  & pct high
          3) change high & pct low
          4) change high & pct high

    Brand-only tables are removed (only row tables are displayed).
    """

    dfs = {"df_all": df_all, "df_a": df_a, "df_b": df_b, "df_c": df_c, "df_d": df_d}

    # --- widgets
    df_name = widgets.Dropdown(
        options=list(dfs.keys()),
        value="df_all" if "df_all" in dfs else list(dfs.keys())[0],
        description="DF",
        layout=widgets.Layout(width="260px")
    )

    category_dd = widgets.Dropdown(
        options=[],
        description="Category",
        layout=widgets.Layout(width="260px")
    )

    bins_change = widgets.IntSlider(
        value=default_bins, min=5, max=200, step=1, description="Bins (chg)",
        layout=widgets.Layout(width="520px")
    )
    bins_pct = widgets.IntSlider(
        value=default_bins, min=5, max=200, step=1, description="Bins (pct)",
        layout=widgets.Layout(width="520px")
    )

    thr_change = widgets.FloatRangeSlider(
        value=list(default_thr_change), min=0.0, max=1.0, step=0.01,
        description="Change Q", readout_format=".2f", continuous_update=True,
        layout=widgets.Layout(width="520px")
    )
    thr_pct = widgets.FloatRangeSlider(
        value=list(default_thr_pct), min=0.0, max=1.0, step=0.01,
        description="Pct Q", readout_format=".2f", continuous_update=True,
        layout=widgets.Layout(width="520px")
    )

    row_sort = widgets.Dropdown(
        options=[("Keep current order", "keep"), ("Sort by extremeness", "extreme")],
        value="keep",
        description="Row sort",
        layout=widgets.Layout(width="260px")
    )

    summary = widgets.HTML()
    out = widgets.Output()

    def _sanitize(q0, q1):
        q0 = max(0.0, min(float(q0), 0.999))
        q1 = max(0.001, min(float(q1), 1.0))
        if q0 >= q1:
            q0 = max(0.0, q1 - 0.01)
        return q0, q1

    def _available_categories(cur_df: pd.DataFrame):
        cols = list(cur_df.columns)
        prefixes = sorted({c.split("_")[0] for c in cols if c.endswith("_change")})
        prefixes = [p for p in prefixes if f"{p}_pct" in cols]
        pretty = {"water": "water", "hwater": "hwater", "elect": "elect", "heat": "heat"}
        return [(pretty.get(p, p), p) for p in prefixes]

    def _display_cols_for_prefix(cur_df: pd.DataFrame, prefix: str, change_col: str, pct_col: str):
        df_cols = list(cur_df.columns)

        # base: building + brand early
        final = []
        for c in [building_col, brand_col]:
            if c in df_cols and c not in final:
                final.append(c)

        # add first few "context" cols (excluding brand/building)
        base = [c for c in df_cols[:5] if c not in (brand_col, building_col)]
        for c in base:
            if c in df_cols and c not in final:
                final.append(c)

        usage_candidates = [f"{prefix}_usage_m3", f"{prefix}_usage_kw", f"{prefix}_usage_m3_mwh"]
        usage_col = next((c for c in usage_candidates if c in df_cols), None)

        # change/pct before prev/current (your requirement)
        cat_block = []
        for c in [change_col, pct_col, f"{prefix}_previous", f"{prefix}_current", usage_col]:
            if c and c in df_cols:
                cat_block.append(c)

        for c in cat_block:
            if c not in final:
                final.append(c)

        return final

    def _update_categories(*_):
        cur_df = dfs[df_name.value]
        for must in (brand_col, building_col):
            if must not in cur_df.columns:
                raise ValueError(f"Column '{must}' not found in selected df ({df_name.value})")

        opts = _available_categories(cur_df)
        if not opts:
            category_dd.options = []
            category_dd.value = None
            return

        prev = category_dd.value
        category_dd.options = opts
        valid_values = [v for _, v in opts]
        category_dd.value = prev if prev in valid_values else valid_values[0]

    def _section_header(text: str):
        return widgets.HTML(
            f"<div style='margin:14px 0 6px 0; padding:8px 10px; border:1px solid #ddd; "
            f"border-radius:8px; font-weight:700; font-size:15px;'>{text}</div>"
        )

    def _table_box(title_html: str, df_to_show: pd.DataFrame):
        box_out = widgets.Output()
        with box_out:
            display(df_to_show)
        return widgets.VBox([widgets.HTML(title_html), box_out], layout=widgets.Layout(width="50%"))

    def _maybe_sort(df_part: pd.DataFrame, sort_cols, ascending):
        if row_sort.value != "extreme" or df_part.empty:
            return df_part
        return df_part.sort_values(sort_cols, ascending=ascending)

    def redraw(*_):
        with out:
            out.clear_output(wait=True)

            cur_df = dfs[df_name.value].copy()
            if category_dd.value is None:
                display(widgets.HTML("<b style='color:#b00'>No category available in this df.</b>"))
                return

            prefix = category_dd.value
            change_col = f"{prefix}_change"
            pct_col = f"{prefix}_pct"

            # numeric coercion
            cur_df[change_col] = pd.to_numeric(cur_df[change_col], errors="coerce")
            cur_df[pct_col] = pd.to_numeric(cur_df[pct_col], errors="coerce")

            both = cur_df[[brand_col, building_col, change_col, pct_col]].dropna()
            if both.empty:
                display(widgets.HTML("<b style='color:#b00'>No rows with both numeric change & pct.</b>"))
                return

            s_change = both[change_col]
            s_pct = both[pct_col]

            q0c, q1c = _sanitize(*thr_change.value)
            q0p, q1p = _sanitize(*thr_pct.value)

            lo_c, hi_c = float(s_change.quantile(q0c)), float(s_change.quantile(q1c))
            lo_p, hi_p = float(s_pct.quantile(q0p)), float(s_pct.quantile(q1p))

            summary.value = (
                f"<div style='font-size:12px;color:#555'>"
                f"<b>{df_name.value}</b> | <b>{prefix}</b> | "
                f"Change ({change_col}) q[{q0c:.2f},{q1c:.2f}] → <b>[{lo_c:.6g}, {hi_c:.6g}]</b> &nbsp; | &nbsp; "
                f"Pct ({pct_col}) q[{q0p:.2f},{q1p:.2f}] → <b>[{lo_p:.6g}, {hi_p:.6g}]</b>"
                f"</div>"
            )

            # ---------- PLOTS ----------
            fig, axes = plt.subplots(1, 2, figsize=(14, 5))

            axes[0].hist(s_change, bins=bins_change.value, edgecolor="black", alpha=0.7)
            xmin_c, xmax_c = float(s_change.min()), float(s_change.max())
            if lo_c > xmin_c:
                axes[0].axvspan(xmin_c, lo_c, alpha=0.25, zorder=3)
            if hi_c < xmax_c:
                axes[0].axvspan(hi_c, xmax_c, alpha=0.25, zorder=3)
            axes[0].axvline(lo_c, linestyle="--", linewidth=2, zorder=4)
            axes[0].axvline(hi_c, linestyle="--", linewidth=2, zorder=4)
            axes[0].set_title(f"Change: {change_col}")

            axes[1].hist(s_pct, bins=bins_pct.value, edgecolor="black", alpha=0.7)
            xmin_p, xmax_p = float(s_pct.min()), float(s_pct.max())
            if lo_p > xmin_p:
                axes[1].axvspan(xmin_p, lo_p, alpha=0.25, zorder=3)
            if hi_p < xmax_p:
                axes[1].axvspan(hi_p, xmax_p, alpha=0.25, zorder=3)
            axes[1].axvline(lo_p, linestyle="--", linewidth=2, zorder=4)
            axes[1].axvline(hi_p, linestyle="--", linewidth=2, zorder=4)
            axes[1].set_title(f"Pct: {pct_col}")

            plt.tight_layout()
            plt.show()

            cols_to_show = _display_cols_for_prefix(cur_df, prefix, change_col, pct_col)

            # ---------- CHANGE TAILS (ROWS) ----------
            display(_section_header("Change tails"))

            chg_low_mask = cur_df[change_col] < lo_c
            chg_high_mask = cur_df[change_col] > hi_c
            chg_low_df = cur_df.loc[chg_low_mask].copy()
            chg_high_df = cur_df.loc[chg_high_mask].copy()

            if row_sort.value == "extreme":
                chg_low_df = chg_low_df.sort_values(change_col, ascending=True)
                chg_high_df = chg_high_df.sort_values(change_col, ascending=False)

            display(widgets.HBox([
                _table_box(f"<b>Change LOW</b> (rows={len(chg_low_df)})", chg_low_df[cols_to_show].head(max_rows_show)),
                _table_box(f"<b>Change HIGH</b> (rows={len(chg_high_df)})", chg_high_df[cols_to_show].head(max_rows_show)),
            ]))

            # ---------- PCT TAILS (ROWS) ----------
            display(_section_header("Pct tails"))

            pct_low_mask = cur_df[pct_col] < lo_p
            pct_high_mask = cur_df[pct_col] > hi_p
            pct_low_df = cur_df.loc[pct_low_mask].copy()
            pct_high_df = cur_df.loc[pct_high_mask].copy()

            if row_sort.value == "extreme":
                pct_low_df = pct_low_df.sort_values(pct_col, ascending=True)
                pct_high_df = pct_high_df.sort_values(pct_col, ascending=False)

            display(widgets.HBox([
                _table_box(f"<b>Pct LOW</b> (rows={len(pct_low_df)})", pct_low_df[cols_to_show].head(max_rows_show)),
                _table_box(f"<b>Pct HIGH</b> (rows={len(pct_high_df)})", pct_high_df[cols_to_show].head(max_rows_show)),
            ]))

            # ---------- OVERLAP: 4 QUADRANTS (ROWS) ----------
            display(_section_header("Overlap (Change × Pct quadrants)"))

            chg_low = chg_low_mask
            chg_high = chg_high_mask
            pct_low = pct_low_mask
            pct_high = pct_high_mask

            q_LL = cur_df.loc[chg_low & pct_low].copy()
            q_LH = cur_df.loc[chg_low & pct_high].copy()
            q_HL = cur_df.loc[chg_high & pct_low].copy()
            q_HH = cur_df.loc[chg_high & pct_high].copy()

            if row_sort.value == "extreme":
                q_LL = _maybe_sort(q_LL, [pct_col, change_col], [True, True])
                q_LH = _maybe_sort(q_LH, [pct_col, change_col], [False, True])
                q_HL = _maybe_sort(q_HL, [pct_col, change_col], [True, False])
                q_HH = _maybe_sort(q_HH, [pct_col, change_col], [False, False])

            display(widgets.VBox([
                widgets.HBox([
                    _table_box(f"<b>Change LOW · Pct LOW</b> (rows={len(q_LL)})", q_LL[cols_to_show].head(max_rows_show)),
                    _table_box(f"<b>Change LOW · Pct HIGH</b> (rows={len(q_LH)})", q_LH[cols_to_show].head(max_rows_show)),
                ]),
                widgets.HBox([
                    _table_box(f"<b>Change HIGH · Pct LOW</b> (rows={len(q_HL)})", q_HL[cols_to_show].head(max_rows_show)),
                    _table_box(f"<b>Change HIGH · Pct HIGH</b> (rows={len(q_HH)})", q_HH[cols_to_show].head(max_rows_show)),
                ]),
            ]))

    # wire up
    df_name.observe(_update_categories, names="value")
    for w in (category_dd, bins_change, bins_pct, thr_change, thr_pct, row_sort):
        w.observe(redraw, names="value")

    # init
    _update_categories()

    ui = widgets.VBox([
        widgets.HTML("<h3 style='margin:0 0 6px 0;'>DF + Category Outlier Dashboard</h3>"),
        widgets.HBox([df_name, category_dd, row_sort]),
        widgets.HBox([bins_change, thr_change]),
        widgets.HBox([bins_pct, thr_pct]),
        summary,
        out
    ])
    display(ui)
    redraw()


# -------- Example call (your case) --------
interactive_overlap_dashboard_v2(
    df_all=df_all, df_a=df_a, df_b=df_b, df_c=df_c, df_d=df_d,
    building_col="building", brand_col="brand"
)
```

```python
import numpy as np
import plotly.graph_objects as go

def hist2D_heatmap(
    df,
    col_1,
    col_2,
    bins=30,
    title=None,
    xaxis_title=None,
    yaxis_title=None,
    log_scale=False
):
    # clean numeric + drop NaNs
    x = np.asarray(df[col_1], dtype=float)
    y = np.asarray(df[col_2], dtype=float)
    m = np.isfinite(x) & np.isfinite(y)
    x = x[m]
    y = y[m]

    hist, xedges, yedges = np.histogram2d(x, y, bins=bins)

    # bin centers
    xcenters = 0.5 * (xedges[:-1] + xedges[1:])
    ycenters = 0.5 * (yedges[:-1] + yedges[1:])

    z = hist.T
    if log_scale:
        z = np.log1p(z)
        colorbar_title = "count (log scale)"
    else:
        colorbar_title = "count"

    fig = go.Figure(
        data=go.Heatmap(
            x=xcenters,
            y=ycenters,
            z=z,
            colorscale="Blues",
            colorbar=dict(title=colorbar_title),
        )
    )

    fig.add_vline(x=0, line_width=1, line_color="black")
    fig.add_hline(y=0, line_width=1, line_color="black")

    fig.update_layout(
        title=title or f"{col_1} vs {col_2}",
        xaxis=dict(title=xaxis_title or col_1),
        yaxis=dict(title=yaxis_title or col_2),
    )

    fig.show()
```

```python
col_1, col_2 = "water_change", "water_pct"

hist2D_heatmap(
    df_all,
    col_1,
    col_2,
    bins=30,
    title="Water change vs pct",
    log_scale=False
)
```

```python
import numpy as np
import plotly.graph_objects as go

def hist3D_interactive(
    df,
    col_1,
    col_2,
    bins=30,
    log_scale=False,
    title=None,
    xaxis_title=None,
    yaxis_title=None,
    zaxis_title=None,
):
    # clean numeric + drop NaNs
    x = np.asarray(df[col_1], dtype=float)
    y = np.asarray(df[col_2], dtype=float)
    m = np.isfinite(x) & np.isfinite(y)
    x = x[m]
    y = y[m]

    hist, xedges, yedges = np.histogram2d(x, y, bins=bins)

    xcenters = 0.5 * (xedges[:-1] + xedges[1:])
    ycenters = 0.5 * (yedges[:-1] + yedges[1:])

    # mesh for surface
    xpos, ypos = np.meshgrid(xcenters, ycenters)
    zpos = hist.T

    if log_scale:
        zpos = np.log1p(zpos)

    # # symmetric extents
    # max_x = float(np.max(np.abs(xcenters))) if len(xcenters) else 1.0
    # max_y = float(np.max(np.abs(ycenters))) if len(ycenters) else 1.0

    # robust axis limits (percentile-based)
    x_lo, x_hi = np.percentile(x, [1, 99])
    y_lo, y_hi = np.percentile(y, [1, 99])

    fig = go.Figure(
        data=[go.Surface(
            x=xpos,
            y=ypos,
            z=zpos,
            colorscale="Blues",
            contours=dict(z=dict(show=True))
        )]
    )

    fig.update_layout(
        title=title or f"{col_1} vs {col_2} (3D histogram)",
        # scene=dict(
        #     xaxis=dict(title=xaxis_title or col_1, range=[-max_x, max_x], zeroline=True),
        #     yaxis=dict(title=yaxis_title or col_2, range=[-max_y, max_y], zeroline=True),
        #     zaxis=dict(title=zaxis_title or ("log(count+1)" if log_scale else "count")),
        # ),
        scene=dict(
            xaxis=dict(title=xaxis_title or col_1, range=[x_lo, x_hi]),
            yaxis=dict(title=yaxis_title or col_2, range=[y_lo, y_hi]),
            zaxis=dict(title=zaxis_title or ("log(count+1)" if log_scale else "count")),
    ),
        margin=dict(l=0, r=0, t=40, b=0),
    )

    fig.show()
```

```python
hist3D_interactive(
    df_all,
    "water_change",
    "water_pct",
    bins=30,
    log_scale=True,
    title="Water change vs pct (3D)"
)
```
